# Prepare datasets

```python
import pandas as pd

df = pd.read_csv("editing/AudioEditingCode/MedleyMDPrompts/captions_gpt5.csv")
target_size = len(df) // 6
df_hparam = df.groupby("edit", group_keys=False).apply(
    lambda x: x.sample(frac=target_size/len(df), random_state=42)
).drop(columns=["Unnamed: 0"])

df_remaining = df.drop(index=df_hparam.index)
frac_test = 0.5 * len(df) / len(df_remaining)   # scale so it's half of original df
df_test = df_remaining.groupby("edit", group_keys=False).apply(
    lambda x: x.sample(frac=frac_test, random_state=42)
).drop(columns=["Unnamed: 0"])

df_loc = df_remaining.drop(index=df_test.index).drop(columns=["Unnamed: 0"])
```

# Create lower bound set

```python
from pathlib import Path
import shutil

path_audios = Path("/data/SHARE/datasets/MedleyDB_V1").resolve()
```

```python
target_dir = Path("editing/outputs/medleymd/lower_bound_hparam/audios").resolve()
target_dir.mkdir(parents=True, exist_ok=True)

for idx, (_, row) in enumerate(df_hparam.iterrows()):
    filename = row["filename"]
    dirname = filename.split("_MIX")[0]
    path_to_audio = (path_audios / dirname / filename).resolve()
    target_path = (target_dir / f"a{idx}.wav").resolve()
    shutil.copy(path_to_audio, target_path)
```

```python
target_dir = Path("editing/outputs/medleymd/lower_bound_test/audios").resolve()
target_dir.mkdir(parents=True, exist_ok=True)

for idx, (_, row) in enumerate(df_test.iterrows()):
    filename = row["filename"]
    dirname = filename.split("_MIX")[0]
    path_to_audio = (path_audios / dirname / filename).resolve()
    target_path = (target_dir / f"a{idx}.wav").resolve()
    shutil.copy(path_to_audio, target_path)
```

```python
target_dir = Path("editing/outputs/medleymd/lower_bound_loc/audios").resolve()
target_dir.mkdir(parents=True, exist_ok=True)

for idx, (_, row) in enumerate(df_loc.iterrows()):
    filename = row["filename"]
    dirname = filename.split("_MIX")[0]
    path_to_audio = (path_audios / dirname / filename).resolve()
    target_path = (target_dir / f"a{idx}.wav").resolve()
    shutil.copy(path_to_audio, target_path)
```
